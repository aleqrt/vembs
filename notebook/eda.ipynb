```python
def group_classes(race):
    grouping = {
        "White": ['WHITE', 'WHITE - BRAZILIAN', 'WHITE - OTHER EUROPEAN', 'PORTUGUESE', 'WHITE - RUSSIAN', 'WHITE - EASTERN EUROPEAN'],
        "Black": ['BLACK/AFRICAN AMERICAN', 'BLACK/CAPE VERDEAN', 'BLACK/CARIBBEAN ISLAND', 'BLACK/AFRICAN'],
        "Hispanic/Latino": ['HISPANIC/LATINO - GUATEMALAN', 'HISPANIC/LATINO - PUERTO RICAN', 'HISPANIC OR LATINO', 
                            'HISPANIC/LATINO - CUBAN', 'HISPANIC/LATINO - DOMINICAN', 'HISPANIC/LATINO - COLUMBIAN', 
                            'HISPANIC/LATINO - CENTRAL AMERICAN', 'HISPANIC/LATINO - HONDURAN', 'HISPANIC/LATINO - SALVADORAN', 
                            'HISPANIC/LATINO - MEXICAN', 'SOUTH AMERICAN'],
        "Other": ['ASIAN', 'ASIAN - CHINESE', 'ASIAN - SOUTH EAST ASIAN', 'ASIAN - ASIAN INDIAN', 'ASIAN - KOREAN','NATIVE HAWAIIAN OR OTHER PACIFIC ISLANDER', 
                  'AMERICAN INDIAN/ALASKA NATIVE', 'MULTIPLE RACE/ETHNICITY', 'OTHER']#,
        #"Unknown":['UNKNOWN', 'UNABLE TO OBTAIN', 'PATIENT DECLINED TO ANSWER']
    }

    for group, members in grouping.items():
        if race in members:
            return group
    return 'Other'  # Default group if no match found
```

```python
filtered_df['grouped_race'] = filtered_df['race_new'].apply(group_classes)
filtered_df['grouped_race'].value_counts()
```

## Histogram plots and first order statistic

```python
def plot_distributions(dataframe, output_folder):
    # Categorical variables (gender, insurance, race, age_decile)
    categorical_vars = ['gender', 'insurance', 'race', 'grouped_race', 'age_decile']
    
    for var in categorical_vars:
        # Plot pie chart
        plt.figure(figsize=(18, 10))
        counts = dataframe[var].value_counts()
        total = counts.sum()
        filtered_counts = counts[counts / total >= 0.05]
        removed_counts = counts[counts / total < 0.05]

        wedges, texts, autotexts = plt.pie(filtered_counts, labels=filtered_counts.index, autopct='%1.1f%%', startangle=140, wedgeprops={'edgecolor': 'black'})
        if var in ['race', 'grouped_race']:
            plt.title(f'Pie Chart of ethnicity')
        else:
            plt.title(f'Pie Chart of {var}')
        
        # Add note for removed categories if any
        if not removed_counts.empty:
            removed_categories = [str(category) for category in removed_counts.index]
            note = "Categories removed (less than 5% each): " + ", ".join(removed_categories)
            plt.figtext(0.5, -0.1, note, ha="center", fontsize=10)
        
        # Adjust legend
        plt.legend(wedges, filtered_counts.index, loc="center left", bbox_to_anchor=(1, 0.5), fontsize='small')
        plt.setp(autotexts, size=10, weight="bold")
        
        # Save the plot
        plot_path = os.path.join(output_folder, f'{var}_distribution.png')
        plt.savefig(plot_path)
        plt.close()
    
    # Diagnostic variables (binary)
    # diagnostic_vars = [
    #     'Enlarged Cardiomediastinum', 'Cardiomegaly', 'Lung Opacity', 'Lung Lesion', 'Edema', 
    #     'Consolidation', 'Pneumonia', 'Atelectasis', 'Pneumothorax', 'Pleural Effusion', 
    #     'Pleural Other', 'Fracture', 'Support Devices', 'No Finding', 'hospital_expire_flag'
    # ]
    diagnostic_vars = [
        'No Finding'
    ]
    
    for var in diagnostic_vars:
        # Plot pie chart
        plt.figure(figsize=(12, 8))
        counts = dataframe[var].value_counts()
        total = counts.sum()
        filtered_counts = counts[counts / total >= 0.05]
        removed_counts = counts[counts / total < 0.05]

        wedges, texts, autotexts = plt.pie(filtered_counts, labels=filtered_counts.index, autopct='%1.1f%%', startangle=140, wedgeprops={'edgecolor': 'black'})
        plt.title(f'Pie Chart of {var}')
        
        # Add note for removed categories if any
        if not removed_counts.empty:
            removed_categories = [str(category) for category in removed_counts.index]
            note = "Categories removed (less than 5% each): " + ", ".join(removed_categories)
            plt.figtext(0.5, -0.1, note, ha="center", fontsize=10)
        
        # Adjust legend
        plt.legend(wedges, filtered_counts.index, loc="center left", bbox_to_anchor=(1, 0.5), fontsize='small')
        plt.setp(autotexts, size=10, weight="bold")
        
        # Save the plot
        plot_path = os.path.join(output_folder, f'{var}_distribution.png')
        plt.savefig(plot_path)
        plt.close()
```
